# Messy Employees Dataset Cleaning

```python
# importing the libraries
import pandas as pd
import numpy as np
from pandas.core.groupby.grouper import Grouping
from pandas.core.resample import get_resampler_for_grouping
```

```python
# Reading the file using Pandas
df =pd.read_csv('Messy_Employee_dataset.csv')
df.isnull().sum()
```

```python
# Filling the NAN values in the age column by its median
# df['Age'] = df['Age'].interpolate(method='linear')
df['Age'] = df['Age'].fillna(df['Age'].mean(),inplace=True).astype(int)
df
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].mean() , inplace=True).astype(int)
df
```

```python
df.isnull().sum()
```

```python
Grouping = df.groupby(['Employee_ID','First_Name','Last_Name','Department_Region'])['Salary'].sum()
Grouping
```

```python
# Checking if there any infinite Values are Present
df.replace([np.inf, -np.inf], np.nan, inplace=True).sum()
```
